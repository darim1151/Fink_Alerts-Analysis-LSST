# 00 - Fink LSST Access Smoke Test

Checkpoint 0.6 status view for tiny public/no-login Fink LSST REST probes. This notebook does not fetch cutouts, FITS/image data, Data Transfer, Kafka, Livestream, Spark, or login-gated services.

```python
from pathlib import Path
import os
import pandas as pd

if Path.cwd().name == "notebooks":
    os.chdir(Path.cwd().parent)

from fink_lsst.api import FinkApiClient
from fink_lsst.config import load_config
from fink_lsst.storage import read_json

config = load_config("configs/local_smoke_test.yaml")
client = FinkApiClient(config=config)
print(client.base_url)
print(client.build_root_url('/swagger.json'))
```

## Discovery Artifacts

Run `python scripts/discover_fink_lsst_contracts.py` to refresh these files.

```python
paths = {
    "swagger": Path("data/fixtures/fink_lsst_swagger.json"),
    "contract_summary": Path("data/fixtures/fink_lsst_api_contract_summary.json"),
    "schema_probes": Path("outputs/api_contracts/schema_probe_results.json"),
    "endpoint_probes": Path("outputs/api_contracts/endpoint_probe_results.json"),
    "report": Path("outputs/api_contracts/FINK_LSST_API_CONTRACT_REPORT.md"),
}
pd.DataFrame([{"artifact": name, "exists": path.exists(), "path": str(path)} for name, path in paths.items()])
```

## Swagger And Endpoint Contracts

```python
if paths["contract_summary"].exists():
    contract_df = pd.DataFrame(read_json(paths["contract_summary"]))
    display(contract_df[["path", "method", "summary", "required_parameters", "optional_parameters"]])
else:
    print("No contract summary yet. Run scripts/discover_fink_lsst_contracts.py")
```

## Schema Variant Status

```python
if paths["schema_probes"].exists():
    schema_probe_df = pd.DataFrame(read_json(paths["schema_probes"]))
    display(schema_probe_df[["schema_target", "method", "params", "payload", "ok", "status_code", "content_type", "output_file", "response_preview", "error"]])
else:
    print("No schema probe results yet.")
```

## Endpoint Probe Status

```python
if paths["endpoint_probes"].exists():
    endpoint_probe_df = pd.DataFrame(read_json(paths["endpoint_probes"]))
    display(endpoint_probe_df[["name", "method", "url", "ok", "skipped", "status_code", "content_type", "output_file", "skip_reason", "error"]])
else:
    print("No endpoint probe results yet.")
```

## Interpretation

- Confirmed JSON endpoints: rows with `ok == True` and `content_type == application/json`.
- HTML/documentation responses: rows with `parsed_json == False` or `text/html` content type.
- Network failures: rows with DNS, connection, or timeout errors.
- ID-gated endpoints: rows skipped because they require known `diaObjectId` or `diaSourceId`.
- Deferred endpoints: cutout/image paths and login-gated services remain out of scope.
